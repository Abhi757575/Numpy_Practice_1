```python
import numpy as np
```

1D array

```python
list1 = [5, 10, 15, 20]

arr = np.array(list1)

print(arr)
print(arr.shape)
print(arr.ndim)
```

```python
list1 = [ 4, 5, 6, 3, 9, 3]

arr = np.array(list1)
arr = arr.reshape(2,3)

print(arr)
print(arr.shape)
print(arr.ndim)
print(type(arr))
print(arr.size)
```

```python
print(arr.dtype)
```

```python
arr = np.ones((3,3))
print(arr)
```

```python
arr = np.ones((2,4))
print(arr)
```

```python
arr = np.eye(4)
print(arr)
```

```python
arr = np.arange(1, 11)
print(arr)
```

```python
arr = np.array([10, 20, 30, 40, 50, 60])
print(arr[0])
print(arr[-1])
print(arr[2:6])
print(arr[::-1])
```

```python
arr = np.arange(1,13)
print(arr)

arr = arr.reshape(3, 4)
print(arr)
```

```python
arr1 = np.array([1, 2, 3])
arr2 = np.array([5,6,7])

concat = np.concatenate((arr1, arr2))
print(concat)
```

```python
arr3 = np.array([1, 2, 3, 4, 5, 6])
split_arr = np.split(arr3 ,3)
print(split_arr)
```

```python
v_stack_arr = np.vstack((arr1, arr2))
print(v_stack_arr)
```

```python
h_stack_arr = np.hstack((arr1, arr2))
print(h_stack_arr)
```
